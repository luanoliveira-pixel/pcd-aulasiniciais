```python
nomeAluno = input("Qual o seu nome?")

nota1 = float(input("Quanto você tirou na primeira prova? "))
nota2 = float(input("Quanto você tirou na segunda prova? "))
nota3 = float(input("Quanto você tirou na terceira prova? "))

media = (nota1 + nota2 + nota3)/3

print("Nome do aluno: ", nomeAluno)
print("Nota da primeira prova: ", nota1)
print("Nota da segunda prova: ", nota2)
print("Nota da terceira prova: ", nota3)
print("Média do",nomeAluno,":", media)
```
