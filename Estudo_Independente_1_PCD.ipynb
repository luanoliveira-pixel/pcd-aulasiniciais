Desafio 1 – Análise de notas de uma turma
Desenvolva um programa que permita registrar as notas de uma turma de estudantes.
O programa deverá:
* solicitar a quantidade de estudantes;
* armazenar as notas em uma lista;
calcular:
* maior nota;
* menor nota;
* média da turma;
* quantidade de estudantes aprovados (nota ≥ 6,0);
* quantidade de estudantes reprovados.
Os cálculos deverão ser implementados utilizando funções.

```python
def ler_quantidade_estudantes():
    while True:
        entrada = input("Quantos estudantes há na turma? ")
        if entrada.isdigit() and int(entrada) > 0:
            return int(entrada)
        print("Digite um número inteiro positivo.")


def registrar_notas(quantidade):
    notas = []
    for i in range(1, quantidade + 1):
        while True:
            try:
                nota = float(input(f"Nota do estudante {i}: ").replace(",", "."))
                if 0 <= nota <= 10:
                    notas.append(nota)
                    break
                else:
                    print("A nota deve estar entre 0 e 10.")
            except ValueError:
                print("Digite um valor numérico válido.")
    return notas


def calcular_maior_nota(notas):
    return max(notas)


def calcular_menor_nota(notas):
    return min(notas)


def calcular_media(notas):
    return sum(notas) / len(notas)


def contar_aprovados(notas, media_aprovacao=6.0):
    return len([nota for nota in notas if nota >= media_aprovacao])


def contar_reprovados(notas, media_aprovacao=6.0):
    return len([nota for nota in notas if nota < media_aprovacao])


def exibir_relatorio_turma(notas):
    print("\n----- RELATÓRIO DA TURMA -----")
    print(f"Notas registradas : {notas}")
    print(f"Maior nota        : {calcular_maior_nota(notas):.1f}")
    print(f"Menor nota        : {calcular_menor_nota(notas):.1f}")
    print(f"Média da turma    : {calcular_media(notas):.2f}")
    print(f"Aprovados (>=6.0) : {contar_aprovados(notas)}")
    print(f"Reprovados        : {contar_reprovados(notas)}")


def main_desafio1():
    quantidade = ler_quantidade_estudantes()
    notas = registrar_notas(quantidade)
    exibir_relatorio_turma(notas)


main_desafio1()
```

Desafio 2 – Cadastro e consulta de produtos

Desenvolva um programa que permita cadastrar produtos utilizando um dicionário.
Cada produto deverá possuir:
* código;
* nome;
* preço;
* quantidade em estoque.
Ao final, o programa deverá permitir:
* consultar um produto pelo código;
* calcular o valor total do estoque;
* identificar o produto de maior valor unitário.
Organize a solução utilizando funções.


```python
def cadastrar_produtos():
    estoque = []
    quantidade = int(input("Quantos produtos deseja cadastrar? "))
    for i in range(1, quantidade + 1):
        print(f"\n--- Produto {i} ---")
        codigo = input("Código: ").strip()
        nome = input("Nome: ").strip()

        while True:
            try:
                preco = float(input("Preço: ").replace(",", "."))
                break
            except ValueError:
                print("Digite um preço numérico válido.")

        while True:
            entrada_qtd = input("Quantidade em estoque: ")
            if entrada_qtd.isdigit():
                qtd_estoque = int(entrada_qtd)
                break
            print("Digite um número inteiro válido.")

        produto = [codigo, nome, preco, qtd_estoque]
        estoque.append(produto)
    return estoque


def consultar_produto(estoque, codigo):
    for produto in estoque:
        if produto[0] == codigo:
            return produto
    return None


def calcular_valor_total_estoque(estoque):
    total = 0
    for produto in estoque:
        total += produto[2] * produto[3]
    return total


def encontrar_produto_maior_valor(estoque):
    maior = estoque[0]
    for produto in estoque:
        if produto[2] > maior[2]:
            maior = produto
    return maior


def exibir_produto(produto):
    print(f"Código: {produto[0]}")
    print(f"Nome: {produto[1]}")
    print(f"Preço: R$ {produto[2]:.2f}")
    print(f"Quantidade em estoque: {produto[3]}")


def exibir_menu():
    print("\n----- MENU -----")
    print("1 - Consultar produto por código")
    print("2 - Valor total do estoque")
    print("3 - Produto de maior valor unitário")
    print("0 - Sair")
    return input("Escolha uma opção: ").strip()


def main_desafio2():
    estoque = cadastrar_produtos()

    while True:
        opcao = exibir_menu()

        if opcao == "1":
            codigo = input("Digite o código do produto: ").strip()
            produto = consultar_produto(estoque, codigo)
            if produto:
                exibir_produto(produto)
            else:
                print("Produto não encontrado.")

        elif opcao == "2":
            total = calcular_valor_total_estoque(estoque)
            print(f"Valor total do estoque: R$ {total:.2f}")

        elif opcao == "3":
            produto = encontrar_produto_maior_valor(estoque)
            print("Produto de maior valor unitário:")
            exibir_produto(produto)

        elif opcao == "0":
            print("Encerrando o programa.")
            break

        else:
            print("Opção inválida.")


main_desafio2()
```

Desafio 3 – Sistema de Estatísticas

Desenvolva um programa que leia uma sequência de números informados pelo usuário até que seja digitado
um valor de encerramento.
Ao final, o programa deverá apresentar:
* quantidade de valores informados;
* média;
* mediana;
* maior valor;
* menor valor;
* quantidade de números pares;
* quantidade de números ímpares.
Organize todos os cálculos em funções independentes.

```python
def ler_numeros(sentinela="fim"):
    numeros = []
    print(f'Digite os números um por um. Digite "{sentinela}" para encerrar.')
    while True:
        entrada = input("Número (ou 'fim'): ").strip().lower()
        if entrada == sentinela:
            break
        try:
            numeros.append(int(entrada))
        except ValueError:
            print("Entrada inválida. Digite um número inteiro ou 'fim'.")
    return numeros


def calcular_quantidade(numeros):
    return len(numeros)


def calcular_media(numeros):
    return sum(numeros) / len(numeros)


def calcular_mediana(numeros):
    ordenados = sorted(numeros)
    n = len(ordenados)
    meio = n // 2
    if n % 2 == 0:
        return (ordenados[meio - 1] + ordenados[meio]) / 2
    else:
        return ordenados[meio]


def encontrar_maior(numeros):
    return max(numeros)


def encontrar_menor(numeros):
    return min(numeros)


def contar_pares(numeros):
    return len([n for n in numeros if n % 2 == 0])


def contar_impares(numeros):
    return len([n for n in numeros if n % 2 != 0])


def exibir_estatisticas(numeros):
    if not numeros:
        print("Nenhum número foi informado.")
        return
    print("\n----- ESTATÍSTICAS -----")
    print(f"Valores informados : {numeros}")
    print(f"Quantidade         : {calcular_quantidade(numeros)}")
    print(f"Média              : {calcular_media(numeros):.2f}")
    print(f"Mediana            : {calcular_mediana(numeros):.2f}")
    print(f"Maior valor        : {encontrar_maior(numeros)}")
    print(f"Menor valor        : {encontrar_menor(numeros)}")
    print(f"Quantidade de pares  : {contar_pares(numeros)}")
    print(f"Quantidade de ímpares: {contar_impares(numeros)}")


def main_desafio3():
    numeros = ler_numeros()
    exibir_estatisticas(numeros)


main_desafio3()
```
